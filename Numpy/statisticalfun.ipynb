# Stacking and Spliting

stacking 

```python
arrr1 = np.arange(1,13).reshape(3,4)
arrr2 = np.arange(13,25).reshape(3,4)
```

```python
horizontal_array = np.hstack((arrr1,arrr2))
horizontal_array
```

```python
vertical_array = np.vstack((arrr1,arrr2))
vertical_array
```

spliting

```python
horizontal_split = np.hsplit(horizontal_array,2)
horizontal_split
```

```python
vertical_split = np.vsplit(vertical_array,2)
vertical_split
```
