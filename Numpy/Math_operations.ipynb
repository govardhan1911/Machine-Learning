# Vector Op

```python
arr1 = np.arange(1,17).reshape(4,4)
arr2 = np.arange(17,33).reshape(4,4)
```

```python
arr1  + arr2
```

```python
arr1  - arr2
```

```python
arr1  * arr2
```

### Important function in numpy for ML and DS

# max()

```python
array = np.array([[18, 20, 22, 24],
       [26, 28, 30, 32],
       [34, 36, 38, 40],
       [42, 44, 46, 48]])

np.max(array)
```

```python
 np.max(array , axis=1) # row wise max
 
```

```python
np.max(array , axis=0) # col wise max
```

# min()

```python
 np.min(array , axis=1) # row wise max
```

# Sum()

```python
np.sum(array)

np.sum(array,axis=0)
```

# product()

```python
np.prod(array)
```
