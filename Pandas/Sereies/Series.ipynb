```python
import pandas as pd
import numpy as np
```

```python
pd.Series([1, 2, 3, 4, 5] , index = ['a', 'b', 'c', 'd', 'e'], name = 'My Series')
```

# Series using List :

```python
country = ['USA', 'Canada', 'Germany', 'UK', 'France']
pd.Series(country)
```

```python
subjects = ['Math', 'Science', 'English', 'History', 'Art']
subject_marks = [90, 85, 95, 80, 70]
pd.Series(data = subject_marks, index=subjects)
```

# Series Using Numpy array

```python
values = np.arange(10,21)
pd.Series(values)
```

# Series Using Dict

```python
data_items = {'Maths': 90, 'Science': 85, 'English': 95, 'History': 80, 'Art': 70}
pd.Series(data_items)
```

# Series Using custom index

```python
values= [1, 2, 3, 4, 5]
pd.Series(values, index=['a', 'b', 'c', 'd', 'e'], name='My Series')
```

# Series Using scaler value

```python
pd.Series(4,index=['a', 'b', 'c', 'd', 'e'])
```
